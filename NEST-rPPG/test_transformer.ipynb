```python
import torch
from model import BaseNet, arc_net, transformer_net

dummy = torch.randn(2, 3, 9, 128)
base  = BaseNet();        base.eval()
arc   = arc_net();        arc.eval()
trans = transformer_net(); trans.eval()

with torch.no_grad():
    s1, h1, a1 = base(dummy)
    s2, h2, a2 = arc(dummy)
    s3, h3, a3 = trans(dummy)

print("BaseNet      — Sig:", s1.shape, "HR:", h1.shape, "av:", a1.shape)
print("arc_net      — Sig:", s2.shape, "HR:", h2.shape, "av:", a2.shape)
print("transformer  — Sig:", s3.shape, "HR:", h3.shape, "av:", a3.shape)

assert s1.shape == s3.shape and h1.shape == h3.shape and a1.shape == a3.shape
print("✓ transformer_net shapes match BaseNet")
```
